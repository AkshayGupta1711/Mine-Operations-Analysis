```python
import numpy as np

df["explosive_type"] = None
df["burden"] = np.nan
df["spacing"] = np.nan
```

```python
df.columns
```

```python
df.loc[df["powder_factor"] > 0.85, "explosive_type"] = "ANFO"

df.loc[
    (df["powder_factor"] >= 0.75) &
    (df["powder_factor"] <= 0.85),
    "explosive_type"
] = "Emulsion"

df.loc[df["powder_factor"] < 0.75, "explosive_type"] = "Slurry"
```

```python
np.random.seed(42)

df.loc[df["explosive_type"] == "ANFO", "burden"] = np.random.uniform(
    3.4, 4.0,
    len(df[df["explosive_type"] == "ANFO"])
)

df.loc[df["explosive_type"] == "Emulsion", "burden"] = np.random.uniform(
    3.0, 3.5,
    len(df[df["explosive_type"] == "Emulsion"])
)

df.loc[df["explosive_type"] == "Slurry", "burden"] = np.random.uniform(
    2.8, 3.3,
    len(df[df["explosive_type"] == "Slurry"])
)

df["burden"] = df["burden"].round(2)
```

```python
df.loc[df["explosive_type"] == "ANFO", "spacing"] = np.random.uniform(
    4.2, 5.0,
    len(df[df["explosive_type"] == "ANFO"])
)

df.loc[df["explosive_type"] == "Emulsion", "spacing"] = np.random.uniform(
    3.8, 4.5,
    len(df[df["explosive_type"] == "Emulsion"])
)

df.loc[df["explosive_type"] == "Slurry", "spacing"] = np.random.uniform(
    3.5, 4.1,
    len(df[df["explosive_type"] == "Slurry"])
)

df["spacing"] = df["spacing"].round(2)
```
